```python
from pymongo import MongoClient
import pandas as pd

client = MongoClient(host='localhost', port=27017)

db = client['news_db']

collection = db.news_info

df = pd.DataFrame(list(collection.find()))

df.drop('_id', axis=1, inplace=True)

df
```

```python
df.drop_duplicates(inplace=True)
df
```
